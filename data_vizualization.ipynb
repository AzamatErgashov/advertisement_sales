```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
import numpy as np 
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/anvarnarz/praktikum_datasets/main/reklama_sales.csv')
# df.to_csv('reklama_sales.csv')
df.head(20)
```

```python
sns.histplot(data=df)
plt.show()
```

```python
sns.jointplot(data=df)
plt.show()
```

```python
sns.jointplot(data=df, x='TV', y='Sales')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.regplot(data=df, x='TV', y='Sales', line_kws={"color": "red"})
plt.show()
```

```python
sns.barplot(data=df)
plt.show()
```

```python
data_corr = df.corr()
```

```python
data_corr
```

```python
sns.heatmap(data=data_corr, cmap="YlGnBu", linewidths=100, square=True)
```

```python
sns.displot(data=df, x='TV',y='Sales', kind='kde', height=5, aspect=2)
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.ecdfplot(data=df, x="TV")
sns.ecdfplot(data=df, x="Sales")
sns.ecdfplot(data=df, x="Radio")
sns.ecdfplot(data=df, x="Newspaper")

plt.grid()
plt.show()
```

```python
plt.figure(figsize=(16,6))
plt.xticks(rotation=90)# bu xdagi har bir chiziqdagi datani ogirib beradi
sns.lineplot(data=df)
plt.show()
```
