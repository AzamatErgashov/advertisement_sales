```python
import sklearn
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/anvarnarz/praktikum_datasets/main/reklama_sales.csv')
df.head(10)
```

# Machina learning

## simple regressor

```python
x = df['TV'].to_numpy()
y = df['Sales'].to_numpy()
```

```python
class simple_linear_regressor():
    def __init__(self, x_test):
        self.x_test = x_test

    def to_reply(self):
        x = df['TV'].to_numpy()
        y = df['Sales'].to_numpy()
        Xmean = np.mean(x)
        ymean = np.mean(y)
        theta1 = sum((x-Xmean)*(y-ymean))/sum((x-Xmean)**2)
        theta0 = ymean - theta1*Xmean
        y_predict = theta0 + theta1 * self.x_test
        return y_predict
x_test = df.sample(100, random_state=42)['TV'].to_numpy()
simple_regressor = simple_linear_regressor(x_test)
```

```python
print(x_test)
simple_regressor.to_reply()
```

```python
sns.ecdfplot(data=x_test)
sns.ecdfplot(data=simple_regressor.to_reply())
plt.grid()
plt.show()
```
